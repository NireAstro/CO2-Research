```python
import astropy.io.ascii
import matplotlib.pyplot as plt

def read_and_filter_data(filename, min_wav=1.6, max_wav=1.614):
    data = astropy.io.ascii.read(filename)
    filtered_data = data[(data['col1'] >= min_wav) & (data['col1'] <= max_wav)]
    return filtered_data['col1'], filtered_data['col3']

wav_1, co2_1 = read_and_filter_data("psg_trn_10.txt")
wav_095, co2_095 = read_and_filter_data("psg_trn_095.txt")
wav_09, co2_09 = read_and_filter_data("psg_trn_09.txt")
wav_085, co2_085 = read_and_filter_data("psg_trn_085.txt")
wav_08, co2_08 = read_and_filter_data("psg_trn_08.txt")
wav_075, co2_075 = read_and_filter_data("psg_trn_075.txt")
wav_07, co2_07 = read_and_filter_data("psg_trn_07.txt")
wav_065, co2_065 = read_and_filter_data("psg_trn_065.txt")
wav_06, co2_06 = read_and_filter_data("psg_trn_06.txt")
wav_055, co2_055 = read_and_filter_data("psg_trn_055.txt")
wav_05, co2_05 = read_and_filter_data("psg_trn_05.txt")
wav_045, co2_045 = read_and_filter_data("psg_trn_045.txt")
wav_04, co2_04 = read_and_filter_data("psg_trn_04.txt")

plt.figure(figsize=(16, 5))

plt.plot(wav_1, co2_1, label="425 ppm")
plt.plot(wav_095, co2_095, label="403.75 ppm")
plt.plot(wav_09, co2_09, label="382.5 ppm")
plt.plot(wav_085, co2_085, label="361.25 ppm")
plt.plot(wav_08, co2_08, label="340 ppm")
plt.plot(wav_075, co2_075, label="318.75 ppm")
plt.plot(wav_07, co2_07, label="297.5 ppm")
plt.plot(wav_065, co2_065, label="276.25 ppm")
plt.plot(wav_06, co2_06, label="255 ppm")
plt.plot(wav_055, co2_055, label="233.75 ppm")
plt.plot(wav_05, co2_05, label="212.5 ppm")
plt.plot(wav_045, co2_045, label="191.25 ppm")
plt.plot(wav_04, co2_04, label="170 ppm")

plt.xlabel("Wavelength (um)")
plt.ylabel("Absorbance")
plt.xlim(1.60, 1.614)

plt.title("CO2 Absorption Spectra")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

ppms = [425, 403.75, 382.5, 361.25, 340, 318.75, 297.5, 276.25, 255, 233.75, 212.5, 191.25, 170]
ppm_ = [ppm / 425 for ppm in ppms]
areas = [area_d1, area_d095, area_d09, area_d085, area_d08, area_d075, area_d07, area_d065, area_d06, area_d055, area_d05, area_d045, area_d04]/np.max(areas)


values = np.polyfit(ppm_, areas, 1)
fit_line = np.poly1d(values)


fit_x = np.linspace(min(ppm_), max(ppm_), 100)

plt.scatter(ppm_, areas, label='Data')
plt.plot(fit_x, fit_line(fit_x), label='Linear Fit', c='r')
plt.xlabel('CO2 Concentration Factor')
plt.ylabel('Normalized Absorption Area')
plt.title('CO2 Concentration vs. Absorption Area')
plt.legend()
plt.show()
```
